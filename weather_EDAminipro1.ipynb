```python
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

plt.figure(figsize=(12, 6))
missing.plot(kind="bar")
plt.title("Missing Values in Weather Australia Dataset")
plt.ylabel("Number of Missing Values")
plt.show()
```

```python
df_clean = df.copy()
df_clean = df_clean.dropna(subset=["RainTomorrow"])
```

```python
df_clean["RainTomorrow"].isnull().sum()
```

```python
#handling null values in maxTemp and Rainfall
#numeric columns can be filled with median
df_clean["MinTemp"] = df_clean["MinTemp"].fillna(
    df_clean["MinTemp"].median()
)

df_clean["MaxTemp"] = df_clean["MaxTemp"].fillna(
    df_clean["MaxTemp"].median()
)

df_clean["Rainfall"] = df_clean["Rainfall"].fillna(
    df_clean["Rainfall"].median()
)
```

```python
#checking
df_clean[["MinTemp", "MaxTemp", "Rainfall"]].isnull().sum()
```

```python
# Fill categorical column with mode - ""frequently occuring value is mode""
df_clean["WindGustDir"] = df_clean["WindGustDir"].fillna(
    df_clean["WindGustDir"].mode()[0]
)
```

```python
#checking
df_clean["WindGustDir"].isnull().sum()
```

```python
# Fill categorical columns with mode
cat_cols = ["WindDir9am", "WindDir3pm", "RainToday"]

for col in cat_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])
```

```python
df_clean[cat_cols].isnull().sum()
```

```python
#checking null values
df_clean.isnull().sum().sort_values(ascending=False)
```

```python
#numeric columns are filling with median value
num_cols = [
    "Sunshine", "Evaporation",
    "Cloud9am", "Cloud3pm",
    "Pressure9am", "Pressure3pm",
    "WindGustSpeed", "Humidity3pm",
    "Temp3pm", "WindSpeed3pm",
    "Humidity9am", "WindSpeed9am", "Temp9am"
]

for col in num_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())
```

```python
df_clean.isnull().sum().sort_values(ascending=False)
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df_clean["MaxTemp"], bins=30)
plt.xlabel("Maximum Temperature (°C)")
plt.ylabel("Frequency")
plt.title("Distribution of Maximum Temperature")
plt.show()
```

```python
#observations:
#Most frequent range: Approximately 18°C to 24°C.
#The maximum temperature distribution has a single peak around 21°C to 22°C.
#Most observations fall between approximately 15°C and 30°C, while extreme temperatures occur less frequently.
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df_clean["MinTemp"], bins=30)

plt.xlabel("Minimum Temperature (°C)")
plt.ylabel("Frequency")
plt.title("Distribution of Minimum Temperature")

plt.show()
```

```python
#observation:
#The minimum temperature distribution has a single peak around 11°C to 13°C.
#Most observations fall between approximately 5°C and 20°C, while very low and high minimum temperatures occur less frequently.
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df_clean["Rainfall"], bins=30)

plt.xlabel("Rainfall (mm)")
plt.ylabel("Frequency")
plt.title("Distribution of Rainfall")

plt.show()
```

```python
#observation:
#Most observations have low rainfall, while a small number of days experience heavy rainfall.
#Therefore, the median may be more representative than the mean when summarizing rainfall.
```
